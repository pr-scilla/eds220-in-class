### Check-in: 

1. The integer number -999 is often used to represent missing values. Create a pandas.Series named s with four integer values, two of which are -999. The index of this series should be the letters A through D.
2. In the [pandas.Series documentation](https://pandas.pydata.org/docs/reference/api/pandas.Series.html), look for the method mask(). Use this method to update the series s so that the -999 values are replaced by NA values. HINT: check the first example in the method’s documentation.

```python
import pandas as pd
import numpy as np
```

```python
# Check-in 1

s = pd.Series([6, 7, -999, -999], index = ['A', 'B', 'C', 'D'])
s
```

```python
# Check-in 2

s = s.mask(s == -999)
s
```
