## Import libraries

```python
import os

from pandas.api.types import is_string_dtype, is_numeric_dtype

import matplotlib.pyplot as plt

import pandas as pd
import geopandas as gpd

# Display all columns when looking at dataframes
pd.set_option('display.max_columns', None)
```

## Import power plants data

```python
URL = 'https://raw.githubusercontent.com/carmengg/eds-220-book/refs/heads/main/data/power_plants_epsg4269.csv'
power_plants = pd.read_csv(URL)

power_plants.head(3)
```

## Import US states shapefile

```python
fp = os.path.join('data', 'tl_2026_us_state', 'tl_2026_us_state.shp')
states = gpd.read_file(fp)

states.head(3)
```

## Create geodataframe from CSV

Known from the metadata that the CRS of the power plants data has the code EPSG:4269

```python
# Create points from latitude and longitude
points = gpd.points_from_xy(power_plants['longitude'],
                   power_plants['latitude'])

# Create geodataframe
power_plants = gpd.GeoDataFrame(power_plants, # Data
                                geometry = points,
                                crs = 'EPSG:4269')
```

```python
# Check that power_plants has been updated to include geometry
power_plants.geometry
```

```python
power_plants.plot()
```

## for loops

```python
print(states.crs)
states.plot()
```

```python
states.head(3)
```

```python
# Simplify column names
states.columns = states.columns.str.lower()
states.columns
```

```python
# Check data types of codes columns

print(f"region is string dtype? {is_string_dtype(states['region'])}")
print(f"statefp is string dtype? {is_string_dtype(states['statefp'])}")
print(f"division is string dtype? {is_string_dtype(states['division'])}")
```

```python
# Check data types using for loop instead
code_cols = ['region', 'division', 'statefp']

# Check whether the code columns are strings

for column in code_cols:
    print(f"{column} is string dtype? {is_string_dtype(states[column])}")
```

## `assert`

We want to update the data type of columns from strings to integers, and we want to check it worked without printing things.

```python
# Update code columns into integers
for column in code_cols:
    states[column] = states[column].astype('int')
    assert is_numeric_dtype(states[column]) # because the assertion did not result in any errors, it means this value is true.
```

How does `assert` work?

```python
2 + 2 == 4
```

```python
# Does nothing if the statement is True
assert 2 + 2 == 4
```

```python
# Get assertion error if statement is False
assert 2 + 2 == 3
```

## Subset data

```python
# Column names for columns with codes
code_cols
```

```python
for column in code_cols:
    print(f"{column}:{sorted(states[column].unique())}")
```

```python
states[states['region'] == 9]['name']
```

```python
# Select contiguous US states
contiguous = states[(states['region'] != 9) & (~states['statefp'].isin([2, 15]))] # Tilde reads as not. Select things that are not in 2 and 15
```

```python
# Select the data in power_plants dataframe for the contiguous United States
power_plants = power_plants[(~power_plants['state'].isin(['Alaska', 'Hawaii','Puerto Rico']))]
```

```python
# Confirm that CRS's match
assert contiguous.crs == power_plants.crs
```

```python
# Initialize empty figure
fig, ax = plt.subplots(figsize=(9, 5)) # Update figure size

# Remove the axis for a cleaner map
ax.axis('off')

# Title for the plot
ax.set_title('Operable electric generating plants in the contiguous United States', 
              fontsize=15)

# Add states
contiguous.plot(ax=ax,
               color='none',
               edgecolor='#362312')

# Add electric power plants colored by energy source
power_plants.plot(ax=ax, 
                  column='primsource',
                  legend=True,
                  markersize=4,
                  cmap='tab20',
                  alpha=0.5,
                  legend_kwds={
                      'title': 'Primary energy source',
                      'title_fontsize': 'small',
                      'fontsize': 'small',
                      'loc': 'upper left',
                      'bbox_to_anchor': (0, 0),
                      'ncols': 6  
                  })
                  
plt.show()
```

## `for` with `zip`

Often, we need to simultaneously iterate over two lists. 
`zip()` lets you combine two or more lists to iterate over them

```python
# Iterate over a single list
numbers = [1, 2, 3]
for x in numbers:
    print(x)

print ('\n') # Print blank line

# Iterate over two lists in pairs using zip()
numbers = [1, 2, 3]
letters = ['a', 'b', 'c']
for x, letter in zip(numbers, letters):
    print(x, letter)
```

```python
fig, axes = plt.subplots(nrows = 1, ncols = 3, figsize = (7,3)) # Figure with one row and three columns

#axes[0].set_title('This is axis 0')
#axes[1].set_title('This is axis 1')
#axes[2].set_title('This is axis 2')
# for loop with zip replaces the above commented code

for ax, i in zip(axes, [0,1,2]):
    ax.set_title(f'This is axis {i}')

plt.show()
```

```python
# Select power plants in California

ca_power_plants = power_plants[power_plants['state'] == 'California']
```

```python
# Create a list named top_sources with CA's top 3 electricity primary sources

print(ca_power_plants['primsource'].value_counts().head(3))

top_sources = ca_power_plants['primsource'].value_counts().index[:3].tolist()
top_sources
```

```python
# Isolate the CA state boundary in a variable named ca_boundary
ca_boundary = contiguous[contiguous['name'] == 'California']
```

```python
# Create a plot

fig, axes = plt.subplots(nrows = 1, ncols = 3, figsize = (6,3)) # Figure with one row and three columns


for ax, source in zip(axes, top_sources):
    ax.set_title(source)

    ca_boundary.plot(ax=ax,
                    color = 'none',
                    edgecolor = 'black')

    subset = ca_power_plants[ca_power_plants['primsource'] == source]
    
    subset.plot(ax=ax,
                markersize = 5,
                alpha = 0.5)
    ax.axis('off')
plt.show()
```
