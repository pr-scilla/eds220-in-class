## Arctic communities by country

We want to calculate the number of arctic communities by country

````python
# Create dataframe with number of communities per country
n_comms = (communities.groupby('country')
                        .size()
                     .reset_index(name='n_communities')) 
                     # make series into a dataframe where count of communities are moved to a column called n_communities
````

Goal: merge the `n_comms` and the `countries` dataframes.

To merge two dataframes they need to have at least one column in common.
Do they have a column in common?

````python
# Check if the two dataframes have columns in common
set(countries.columns).intersection(n_comms.columns)

# set() is a built in object type in python, like a list. Strips everything away except for set object. 
# A set is unordered (list is ordered)
# .intersection works on set() objects.
````

````python
common_columns = set(countries.columns).intersection(n_comms.columns)

# Use if else to check if there are any common columns

if len(common_columns) != 0:
    print(f"Common columns: {common_columns}")
else:
    print("No common columns")
````

Make the previous code into a function called `check_common_columns`, that takes in two dataframes as arguments.

````python
def check_common_columns(df1, df2):

    common_columns = set(df1.columns).intersection(df2.columns)
    
    if len(common_columns) != 0:
        print(f"Common columns: {common_columns}")
    else:
        print("No common columns")
````

## Merge dataframes

We can use the `country_names` dataframe to add the country names into the `n_comms`.

To merge dataframes we can use `pandas.merge()` function

```python
output_df = pd.merge(left_df,
                     right_df,
                     how = type_of_join,
                     on = column_to_join)
```

where

- `output_df` = df resulting from the merge
- `left_df` = df we have on the left side
- `right_df` = df on the right side
- `how` = type of join (inner, left, right, outer)
- `on` = the column to join on, must be present on both dataframes

````python
check_common_columns(n_comms, country_names)
````

````python
n_comms = pd.merge(n_comms, country_names, how = 'left', on = 'country')
n_comms
````

````python
check_common_columns(n_comms, countries)
````

````python
# Join n_comms and countries data using inner join

arctic_countries = pd.merge(countries, n_comms, how = 'inner', on = 'admin')
arctic_countries
````

````python
# Alternative method to above
countries.merge(n_comms, how='inner', on='admin')
````

Note: When merging with a GeoDataFrame, the GeoDataFrame needs to be the left dataframes to keep the geometry.

## Review results

````python
print(f"Countries in n_comms: {len(n_comms)}")
print(f"Countries in arctic_countries: {len(arctic_countries)}")
````

````python
# Countries in n_comms without a match in countries
n_comms[~n_comms['admin'].isin(countries['admin'])]
````

````python
countries[countries['admin'].str.contains('Aland')]
````

````python
# Match the name used in the countries data
n_comms['admin'] = n_comms['admin'].replace('Aland Islands', 'Aland')

# Merge again
arctic_countries = countries.merge(n_comms, how='inner', on='admin')
````

````python
print(f"Countries in arctic_countries: {len(arctic_countries)}")
````

````python
arctic_countries.plot(column = 'n_communities', # indicates which column to color by
                      legend = True)
````

## Reprojecting

````python
# Reproject to a CRS optimized for Arctic region
arctic_countries = arctic_countries.to_crs('epsg:3413')
````

````python
arctic_countries.crs
````

````python
fig, ax = plt.subplots(figsize=(8,6))

ax.axis('off')

arctic_countries.plot(ax=ax,
                      column = 'n_communities', # indicates which column to color by
                      edgecolor='0.6',
                      linewidth=0.5,
                      legend = True,
                      cmap='PuBuGn',
                      legend_kwds={
                          "label":"Number of Communties",
                          "orientation": "horizontal"
                      })

ax.set_title('Communities at or above 55N, 2022\n(at least 10,000 people, 500 in Alaska)')

plt.show()
````
